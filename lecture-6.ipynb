```python
import pandas as pd
import numpy as np
```

```python
import tensorflow as tf
```

```python
X = np.array([
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85]
], dtype=np.float32)

y = np.array([
    0,
    0,
    0,
    1,
    1,
    1
], dtype=np.float32)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1,
                         activation='sigmoid'
                         ,input_shape=(2,))
])
```

```python
model.compile(
    optimizer = 'sgd',
    loss = 'binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(X,y,epochs=100,verbose=1)
```

```python
loss,accuracy = model.evaluate(X,y,verbose=0)
```

```python
print("Loss:", loss)
print("Accuracy:", accuracy)

```

```python
new_student = np.array([
    [6, 80]
], dtype=np.float32)
```

```python
prediction = model.predict(new_student)

print("Probability:", prediction[0][0])
```

```python
if prediction[0][0] >= 0.5:
    print("Pass")
else:
    print("Fail")
```
