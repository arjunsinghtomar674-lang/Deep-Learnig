```python
import pandas as pd
import numpy as np 
```

```python
import tensorflow as tf
```

```python
y_true_reg = np.array([0.8, 0.9, 0.7])
y_pred_reg = np.array([0.7, 0.85, 0.6])
```

# MSE

```python
mse = tf.keras.losses.MeanSquaredError()
```

```python
print(mse(y_true_reg,y_pred_reg).numpy())
```

# MAE

```python
mae = tf.keras.losses.MeanAbsoluteError()
```

```python
print(mae(y_true_reg,y_pred_reg).numpy())
```

# Binary Cross 

```python
y_true_binary = np.array([1, 0, 1, 1])
y_pred_binary = np.array([0.9, 0.2, 0.8, 0.7])

bce = tf.keras.losses.BinaryCrossentropy()

print("Binary Cross Entropy:",
      bce(y_true_binary, y_pred_binary).numpy())
```
