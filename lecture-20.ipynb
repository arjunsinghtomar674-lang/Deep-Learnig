```python
import numpy as np 
import pandas as pd 
import tensorflow as tf
```

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6],
    [7],
    [8]
], dtype=np.float32)

y = np.array([
    0, 0, 0, 0, 1, 1, 1, 1
], dtype=np.float32)
```

```python
def create_model():

    model = tf.keras.Sequential([
        tf.keras.layers.Input(shape=(1,)),
        tf.keras.layers.Dense(8, activation='relu'),
        tf.keras.layers.Dense(1, activation='sigmoid')
    ])

    model.compile(
        optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
        loss='binary_crossentropy',
        metrics=['accuracy']
    )

    return model
```

# Batch Gradient Descent

```python
model_batch = create_model()

history_batch = model_batch.fit(
    X,
    y,
    epochs=10,
    batch_size=len(X),      # Full dataset
    verbose=0
)

print("Batch Gradient Descent")
print("Batch Size:", len(X))
print("Final Loss:", history_batch.history['loss'][-1])
print()
```

# Stochastic 

```python
model_sgd = create_model()

history_sgd = model_sgd.fit(
    X,
    y,
    epochs=10,
    batch_size=1,           # One sample
    verbose=0
)

print("Stochastic Gradient Descent")
print("Batch Size:", 1)
print("Final Loss:", history_sgd.history['loss'][-1])
print()
```

```python
model_mini = create_model()

history_mini = model_mini.fit(
    X,
    y,
    epochs=10,
    batch_size=2,           # Small batch
    verbose=0
)

print("Mini-Batch Gradient Descent")
print("Batch Size:", 2)
print("Final Loss:", history_mini.history['loss'][-1])
```
