```python
import tensorflow as tf
import pandas as pd
```

```python
df = pd.read_csv(r'C:\\Users\\Arjun Singh Tomar\\Downloads\\digit-recognizer\\train.csv')
```

```python
X = df.drop(columns = ['label'],axis = 1)
y = df['label']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size = 0.33,random_state=42
)
```

```python
X_train = X_train[:10000] / 255.0
y_train = y_train[:10000]
X_test = X_test[:2000] / 255.0
y_test = y_test[:2000]
```

```python
# IMPORTANT
X_train = X_train.reshape(-1, 28, 28)
X_test = X_test.reshape(-1, 28, 28)
```

```python
def create_model(neurons=64, deep=False, optimizer='sgd'):

    layers = [
        tf.keras.layers.Flatten(input_shape=(28,28)),
        tf.keras.layers.Dense(neurons, activation='relu')
    ]

    if deep:
        layers.append(
            tf.keras.layers.Dense(64, activation='relu')
        )

    layers.append(
        tf.keras.layers.Dense(10, activation='softmax')
    )

    model = tf.keras.Sequential(layers)

    model.compile(
        optimizer=optimizer,
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy']
    )

    return model
```

```python
model1 = create_model(64, False, 'sgd')
h1 = model1.fit(X_train, y_train, epochs=5, batch_size=32, verbose=0)
```
