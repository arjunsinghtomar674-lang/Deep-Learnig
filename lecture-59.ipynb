# BackPropogation in RNN

```python
import tensorflow as tf 
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense,Embedding,SimpleRNN,Input
import numpy as np
```

```python
X = np.array([
    [[1], [2], [3], [4]],
    [[2], [3], [4], [5]],
    [[3], [4], [5], [6]],
    [[4], [5], [6], [7]]
], dtype=np.float32)

y = np.array([0, 1, 1, 1])
```

```python
model = Sequential()
model.add(Input(shape=(4,1)))
model.add(SimpleRNN(8,activation='tanh'))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
history = model.fit(X,y,epochs = 20,batch_size = 2)
```

```python
prediction = model.predict(X)
```

```python
y_pred = (prediction > 0.5).astype(int).ravel()
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy_score(y,y_pred)
```
