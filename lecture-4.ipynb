# Practical Implementation of Perceptron

```python
import numpy as np 
```

```python
X= np.array([
    [2, 60],
    [3, 65],
    [4, 70],
    [5, 75],
    [6, 80],
    [7, 85]

])
```

```python
y = np.array([0,0,0,1,1,1])
```

```python
weights = np.array([0.5,0.5])
```

```python
bias = 0
```

```python
x = X[0]
```

```python
z = np.dot(x,weights)+bias
```

```python
print(z)
```

```python
def step_function(z):
    if z>=0:
        return 1
    else:
        return 0
```

```python
prediction = step_function(z)
```

```python
print(prediction)
```

```python
actual = y[0]

print("Prediction:", prediction)
print("Actual:", actual)
```
