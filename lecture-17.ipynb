```python
import numpy as np
import pandas as pd
import tensorflow as tf
```

```python
X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5],
    [6]
], dtype=np.float32)

y = np.array([
    [0],
    [0],
    [0],
    [1],
    [1],
    [1]
], dtype=np.float32)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(1,)),
    tf.keras.layers.Dense(8, activation="relu"),
    tf.keras.layers.Dense(4, activation="relu"),
    tf.keras.layers.Dense(1, activation="sigmoid")
])
```

```python
model.compile(
    optimizer = 'sgd',
    loss = 'binary_crossentropy',
    metrics = ['accuracy']
)
```

```python
loss_before, accuracy_before = model.evaluate(
    X,
    y,
    verbose=0
)
```

```python
loss_before
```

```python
accuracy_before
```

```python
model.fit(
    X,
    y,
    epochs=50,
    verbose=0
)
```

```python
loss_after, accuracy_after = model.evaluate(
    X,
    y,
    verbose=0
)
```

```python
loss_after, accuracy_after 
```

```python
prediction = model.predict(X)

print("\nPredicted Probabilities:")
print(prediction)
```
